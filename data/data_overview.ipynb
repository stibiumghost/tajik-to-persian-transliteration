## POETRY
***poetry_texts.csv*** contains texts, matched 'as is', and ***poetry_parts.csv*** contains bayts (couplets), matched after splitting of all texts. While bayts are compared strictly, full texts may have significant differences.

The contents of the two documents overlap only partially.

```python
poetry_texts = pd.read_csv('poetry_texts.csv')
print('Full texts:\n')
poetry_texts.info()

print('\n' + '* '*3 + '\n')

print('Bayts (couplets):\n')
poetry_parts = pd.read_csv('poetry_parts.csv')
poetry_parts.info()
```

***Assorted*** here refers to a collection of smaller forms of poetry, but may include fragments of larger works.

```python
poetry_texts_count = poetry_texts.groupby(['author', 'source'], as_index=False).size().rename(columns= {'size':'texts'})
poetry_parts_count = poetry_parts.groupby(['author', 'source'], as_index=False).size().rename(columns= {'size':'bayts'})
poetry_texts_count.merge(poetry_parts_count, on=('author', 'source'))
```

```python
poetry_texts.sample(1).values
```

```python
poetry_parts.sample(1).values
```

## PROSE
([back to Contents](#contents))

***prose_texts.csv*** contains texts, matched 'as is'. ***prose_parts.csv*** contains sentences, matched after splitting of matched texts. While sentences are compared strictly, full texts may have significant differences.

```python
prose_texts = pd.read_csv('prose_texts.csv')
print('Full texts:\n')
prose_texts.info()

print('\n' + '* '*3 + '\n')

print('Sentences:\n')
prose_parts = pd.read_csv('prose_parts.csv')
prose_parts.info()
```

```python
prose_texts_count = prose_texts.groupby(['author'], as_index=False).size().rename(columns= {'size':'texts'})
prose_parts_count = prose_parts.groupby(['author'], as_index=False).size().rename(columns= {'size':'sents'})
prose_texts_count.merge(prose_parts_count, on=('author'))
```

Since most of the texts are independent articles/posts, their ***source*** is just a number.

```python
prose_texts.sample(5)
```

```python
prose_parts.sample(1).values
```

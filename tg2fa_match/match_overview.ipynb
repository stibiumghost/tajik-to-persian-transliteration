## TEXT COMPARISON
([back to Contents](#contents))

Straightforward "transliteration" of consonants is used to make texts comparable:

```python
from tg2fa_match import tg_cons, fa_cons
for i, j in enumerate(tg_cons):
    print(f'{j} <- \u200e{fa_cons[i]}', end=' '*4)
    if ((i+1) % 13 == 0):
        print()
```

```python
from tg2fa_match import tg, fa    # tg - Tajik, fa - Persian
from tg2fa_match import transcribe, trans_words

text_tg = 'Фориғ зи умеди раҳмату бими азоб'
text_fa = 'فارغ ز امید رحمت و بیم عذاب'
print('Simple:',
      text_tg, transcribe(text_tg, lang=tg),
      text_fa, transcribe(text_fa, lang=fa),
      sep='\n')
print()
print('By word:',
      text_tg, trans_words(text_tg, lang=tg),
      text_fa, trans_words(text_fa, lang=fa),
      sep='\n')
```

Texts will be compared based on the Levenshtein ratio between them:

```python
# %pip install Levenshtein    # Dependency!
from tg2fa_match import ratio

print('The Levanshtein ratio of "chalk" and "cheese" is', round(ratio('chalk', 'cheese'), 2))
print('Identical texts have the Levanshtein ratio of', ratio('egg', 'egg'))
```

### Match texts
Get a list of best-matched pairs between a set of text strings in Tajik and a set of text strings in Persian. (Non-matched text strings are dropped.)

```python
from tg2fa_match import match_texts

texts_tg = ['Тамоми одамон озод ба дунё меоянд ва аз лиҳози манзилату ҳуқуқ бо ҳам баробаранд.',
             'Фориғ зи умеди раҳмату бими азоб,',
             'Озод зи хоку бод в-аз оташу об!']
texts_fa = ['تمام آدمان آزاد به دنیا می‌آیند و از لحاظ منزلت و حقوق با هم برابرند.',
            'فارغ ز امید رحمت و بیم عذاب',
            'همه صاحب عقل و وجدانند، باید نسبت به یکدیگر برادروار مناسبت نمایند.']
match_texts(tg_texts=texts_tg, fa_texts=texts_fa, min_ratio=0.8)
```

### Match words
Get a sequence of matching pairs between a text string in Tajik and a text string in Persian and the corresponding sequences of "transliterations". Non-matched parts of text strings are left as is.

The parameter **window** determines the number of words compared at each step, while parameter **seq_len** sets the maximal possible length of a compound word.

```python
from tg2fa_match import match_words

text_tg = 'Фориғ зи умеди out of place раҳмату бими азоб'
text_fa = 'فارغ ز امید رحمت و بیم عذاب'

match_words(tg_line=text_tg, fa_line=text_fa, window=4)
```

## MATCH MANIPULATION
([back to Contents](#contents))

Easily manipulate matched pairs of text strings:

```python
from tg2fa_match import match_words, ParallelText

text_tg = 'Фориғ зи умеди out of place раҳмату бими азоб'
text_fa = 'فارغ ز امید رحمت و بیم عذاب'

match = ParallelText(
    match_words(tg_line=text_tg, fa_line=text_fa, window=4),
    show=('tg_comp', 'fa_comp', 'ratio'))    #'tg_comp', 'fa_comp' - 'transliterated' strings
match
```

```python
list(match)
```

```python
match.data['inds'] = list(range(len(match)))
match.names += ('inds',)
match.show += ('inds',)
match
```

```python
match.ratio()
```
